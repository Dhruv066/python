```python
import matplotlib.pyplot as plt
x = ['A', 'B', 'C', 'D']
y = [10, 15, 7, 12]
plt.bar(x, y,
        color="red",
        width=0.5,
        edgecolor="black",
        linewidth=1.5
        )
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Simple Bar Chart")
plt.show()
```

```python
plt.barh(x, y)
plt.xlabel("Values")
plt.ylabel("Categories")
plt.title("Horizontal Bar Chart")
plt.show()
```

```python
import numpy as np
x = np.arange(4)
y1 = [10, 15, 7, 12]
y2 = [8, 14, 9, 10]
width = 0.2
plt.bar(x -width/2, y1, width, label='Group 1')
plt.bar(x +width/2, y2, width, label='Group 2')
plt.xticks(x, ['A', 'B', 'C', 'D'])
plt.xlabel("Categories")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
#adding text on the bars
values = [10, 20, 30]
plt.bar(['A', 'B', 'C'], values)
for i, v in enumerate(values):
    plt.text(i, v, str(v))
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]
plt.scatter(x, y,
            s=100,
            c="red",
            marker="*",
            alpha=0.3 #transparency
            )
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Scatter Plot")
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]
plt.scatter(x,y)
plt.xlabel("X values")
plt.ylabel("y labels")
plt.title("Simple Scatter Plot")
plt.grid(True)
plt.show()
```

```python
plt.scatter(x, y,
s=100, # marker size
c='red', # color
marker='o',
alpha= 0.7) # transparency
plt.show()
```

```python
import numpy as np
colors = np.array([10, 20, 30, 40, 50])
plt.scatter(x, y, c=colors, cmap='plasma')
plt.colorbar(label='Color Scale')
plt.show()
```

```python
y2 = [5, 3, 4, 2, 1]
plt.scatter(x, y, label='Set 1')
plt.scatter(x, y2, label='Set 2')
plt.xlabel("X")
plt.ylabel("Y")

plt.legend()
plt.show()
```

```python

x = [1, 2, 3, 4, 5]
y = [5, 7, 6, 8, 7]
plt.scatter(x, y)
# Annotate each point
for i in range(len(x)):
    plt.text(x[i]+0.1, y[i]+0.1, f"({x[i]}, {y[i]})") # small offset
```

# pie chart


```python
sizes = [30, 20, 25, 25]
labels = ['A', 'B', 'C', 'D']
plt.pie(sizes, labels=labels,autopct='%1.1f%%') # adding percentage 
plt.show()
```

### changing colors 

```python
colors = ['skyblue', 'lightgreen', 'pink', 'orange']
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors)
plt.show()
```
