```python
import matplotlib.pyplot as plt
x = ['A', 'B', 'C', 'D']
y = [10, 15, 7, 12]
plt.bar(x, y,
        color="red",
        width=0.5,
        edgecolor="black",
        linewidth=1.5
        )
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Simple Bar Chart")
plt.show()
```

```python
plt.barh(x, y)
plt.xlabel("Values")
plt.ylabel("Categories")
plt.title("Horizontal Bar Chart")
plt.show()
```

```python
import numpy as np
x = np.arange(4)
y1 = [10, 15, 7, 12]
y2 = [8, 14, 9, 10]
width = 0.2
plt.bar(x -width/2, y1, width, label='Group 1')
plt.bar(x +width/2, y2, width, label='Group 2')
plt.xticks(x, ['A', 'B', 'C', 'D'])
plt.xlabel("Categories")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
values = [10, 20, 30]
plt.bar(['A', 'B', 'C'], values)
for i, v in enumerate(values):
    plt.text(i, v, str(v))
```
